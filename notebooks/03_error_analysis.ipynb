# Qualitative error analysis

Select a real prediction file, compare correct and incorrect selections, and examine candidate scores and OOV effects. Keep raw and normalized text visible.

```python
import json

import pandas as pd

from asr_rescoring.config import load_config, resolve_path

config = load_config()
run_id = "20261124_main"
model_id = "order_3_tokens_10000000"
files = sorted(resolve_path(config, "predictions").glob(f"{run_id}_{model_id}_*.jsonl"))
selected_file = files[-1] if files else None  # Change explicitly for the model being analyzed.
predictions = (
    [json.loads(line) for line in selected_file.read_text(encoding="utf-8").splitlines()]
    if selected_file
    else []
)
frame = pd.DataFrame(predictions)
if not frame.empty:
    frame["correct"] = frame.normalized_selected == frame.normalized_reference
print(selected_file.name if selected_file else "No predictions yet.")
```

```python
if not frame.empty:
    display(frame.loc[frame.correct, ["utt_id", "reference", "selected"]].head(10))
    display(frame.loc[~frame.correct, ["utt_id", "reference", "selected"]].head(10))
```

```python
if predictions:
    example = next(
        (row for row in predictions if row["normalized_selected"] != row["normalized_reference"]),
        predictions[0],
    )
    print("Reference:", example["reference"])
    display(pd.DataFrame(example["ranking"]))
```
